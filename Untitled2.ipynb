```python
import pandas as pd
from google.colab import files

# Define data
data = {
    "User_ID": [101, 102, 103, 104, 105],
    "Name": ["Ali", "Maryam", "Reza", "Sara", "Mohammad"],
    "Department": ["Sales", "Tech", "Sales", "HR", "Tech"],
    "Salary": [25000, 32000, 27000, 22000, 35000],
    "Performance_Score": [4.5, 4.8, 3.9, 4.2, 4.9]
}

# Create DataFrame
df = pd.DataFrame(data)

# Save to CSV
file_name = "english_sample_data.csv"
df.to_csv(file_name, index=False, encoding='utf-8-sig')

print(f"File '{file_name}' successfully created!")

# Download file
files.download(file_name)
```
