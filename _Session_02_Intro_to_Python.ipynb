### Code in Fortran - 3. Generation Langauage
(1957 Fortran by John Backus, IBM)



```
program sum_even_numbers                                   ! Start of the program withe the name
    integer :: i, sum                                      ! Define two variables as integer
    sum = 0                                                ! Initialize the sum with value zero

    do i = 1, 10                                           ! Define a loop from one to ten
        if (mod(i, 2) == 0) then                           ! Test if the loop variable is even
            sum = sum + i                                  ! If even then add the number to the sum
        end if                                             ! End of the if condition
    end do                                                 ! End of the loop

    print *, 'Sum of even numbers from 1 to 10:', sum      ! Print the result to the terminal
end program sum_even_numbers                               ! End of the program
```

It is important to have in mind the following rules when creating variables:

# 6. Python Comments

## Goal of comments <br>
Comments are an extremely usefull feature.<br>
As your programs become longer you should add comments within your programs that describe your approach.

## Hash Mark <br>
Any text preceded by the hask mark (pound sign) # is ignored by the Python interpreter.

`````python
# Comment examples

# The constant PI is defind
PI = 3.14

PI = 3.14   #  The constant PI is defind
`````

## Some key rules and conventions for comments



1.   Use Comments Sparingly
  - Comments shoud **explain why**, not **what**
  - Avoid obvious comments
  - Keep comments up to date with code changes



```
# Bad: This is obvious
x = x + 1   # Incrementx by 1

# Good: Explain why
x = x + 1   # Adjust for zero-based index
```




2.   Use Inline Comments for Clarification
  - Place short comments on the same line as the code
  - Separate with a least two spaces



```
total_price = price * quantity  # Calculate total cost
```



3. Use Block Comments for Sections of Code
  - Use explanations spanning multiple lines
  - Start each line with ```# ````



```
# Load and preprocess data:
# - Read from a file
# - Handle missing values
# - Normalize numerical date values
```

4. Use TODOs for Future Work
  - Indicate incomplete tasks or potential improvements



```
# TODO: Optimize this function to reduce time complexity
```

5. Follow PEP8 Guidelines
  - Link: https://peps.python.org/pep-0008/
